```python
# Locate a row in a table using pandas
import pandas as pd

my_data = {
    'Vehicles' : ["Jetour", "Cardillac Escalade V-Series", "Tx v8", "Toyota Sequoia", "Range rover"],
    'Price (USD)' : [2000, 1000, 4000, 5000, 3000],
    'No. of Vehicles' : [1, 2, 1, 1, 1],
    'Rating' : [5, 9, 5, 5, 5]
}

# Create DataFrame from the dictionary

df = pd.DataFrame(my_data)  # Very crucial
print ("Get the second item in the table: ")
print ()
print(df.loc[1])

print ()

df = pd.DataFrame(my_data)  # Very crucial
print ("Get the First and second item from the table: ")
print ()
print(df.loc[[1, 2]])
```
