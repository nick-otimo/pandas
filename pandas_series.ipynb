```python
import pandas as pd
 # Series
 # NB: The S must be uppercase ❤
a = [10, 20, 30, 40, 50]
print (a)
print ()
a_s = pd.Series (a)
print (a_s)
print ()
print (f"The third item in the list is: {a_s[2]}")

# But I now Know how to create the labels myself
print ()
a_s = pd.Series (a, index = ["Mango", "Apple", "Pineapple", "Oranges", "WaterMelon"])
print (a_s)
print (a_s ['Mango'])



```

```python
# Series Using a Dictionary

Calories = {
    "Day1" : 420,
    "Day2" : 320,
    "Day3" : 390
}
print (Calories)
print ()
Calories_s = pd.Series (Calories)
print (Calories_s)
print ()
print (f"Day Two: {Calories_s["Day2"]}")
print ()

# Use the data for only the 1st 2 days
S_Calories = pd. Series (Calories_s, index = ["Day1", "Day2"])
print (S_Calories)
```
